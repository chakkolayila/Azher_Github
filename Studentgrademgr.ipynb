```python
"""Student Grade Manager

A terminal program for storing student marks, assigning grades,
and displaying class statistics.

Duplicate student names are allowed because each student is stored
as a separate record in a list.
"""


def calculate_grade(mark):
    """Return the grade corresponding to the student's mark."""
    if mark >= 90:
        return "A"
    elif mark >= 80:
        return "B"
    elif mark >= 70:
        return "C"
    elif mark >= 60:
        return "D"
    else:
        return "E"


def read_mark():
    """Read and validate a mark between 0 and 100."""
    while True:
        value = input("Enter mark (0-100): ").strip()

        try:
            mark = float(value)

            if mark < 0 or mark > 100:
                print("Invalid mark. Enter a value between 0 and 100.")
                continue

            return mark

        except ValueError:
            print("Invalid input. Enter a valid number.")


def add_student(students):
    """Add a student to the student list."""
    while True:
        name = input("Enter student name: ").strip()

        if name:
            break

        print("Student name cannot be empty.")

    mark = read_mark()
    grade = calculate_grade(mark)

    student = {
        "name": name,
        "mark": mark,
        "grade": grade
    }

    students.append(student)

    print(f"{name} was added successfully.")


def format_mark(mark):
    """Format marks without unnecessary decimal places."""
    if mark.is_integer():
        return str(int(mark))

    return f"{mark:.2f}"


def show_results(students):
    """Display student results and class statistics."""
    if not students:
        print("\nNo student records are available.")
        return

    name_width = max(
        20,
        max(len(student["name"]) for student in students)
    )

    separator = "-" * (name_width + 22)

    print("\nSTUDENT RESULTS")
    print(separator)
    print(f'{"Name":<{name_width}}  {"Mark":>8}  {"Grade":>7}')
    print(separator)

    for student in students:
        name = student["name"]
        mark = format_mark(student["mark"])
        grade = student["grade"]

        print(
            f"{name:<{name_width}}  "
            f"{mark:>8}  "
            f"{grade:>7}"
        )

    print(separator)

    marks = [student["mark"] for student in students]

    class_average = sum(marks) / len(marks)
    highest_mark = max(marks)
    lowest_mark = min(marks)

    print(f"Class average : {class_average:.2f}")
    print(f"Highest mark  : {format_mark(highest_mark)}")
    print(f"Lowest mark   : {format_mark(lowest_mark)}")


def print_menu():
    """Display the main menu."""
    print("\nSTUDENT GRADE MANAGER")
    print("1. Add student")
    print("2. Show results")
    print("3. Quit")


def main():
    """Run the Student Grade Manager."""
    students = []

    while True:
        print_menu()

        choice = input("Choose an option (1-3): ").strip().lower()

        if choice in ("1", "add", "a"):
            add_student(students)

        elif choice in ("2", "show", "s"):
            show_results(students)

        elif choice in ("3", "quit", "q"):
            print("Student Grade Manager closed.")
            break

        else:
            print("Invalid option. Please choose 1, 2, or 3.")


if __name__ == "__main__":
    main()
```
